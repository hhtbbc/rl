# 第 22 节: 实验设计与超参数搜索

## 学习目标
1. 理解 RL 实验的科学方法
2. 学会设计消融实验
3. 掌握超参数搜索策略
4. 理解统计显著性在 RL 中的意义

## 1. 为什么 RL 实验需要特殊设计?

- 高方差: 同一算法不同种子表现差异很大
- 大量超参数: lr, gamma, lambda, epsilon, batch size, n_epochs...
- 计算成本高: 完整训练可能数小时
- 没有标准 benchmark: 环境差异大

## 2. 消融实验设计

### 基本消融 (验证每个组件的贡献)
- PPO: 去掉 clipping -> 退化为 REINFORCE+baseline
- DQN: 去掉 target network -> 观察不稳定性
- DQN: 去掉 experience replay -> 观察 catastrophic forgetting
- A2C: 去掉 entropy bonus -> 观察探索不足

### 设计原则
1. 每次只改变一个变量
2. 控制计算预算 (相同 step 数比较)
3. 至少 3 个随机种子
4. 报告均值和标准差 (或中位数和 IQR)

## 3. 超参数搜索

| 超参数 | 典型范围 | 搜索策略 |
|--------|----------|----------|
| Learning rate | 1e-5 ~ 1e-2 | Log-uniform |
| gamma (折扣) | 0.9 ~ 0.999 | Linear |
| lambda (GAE) | 0.8 ~ 1.0 | Linear |
| epsilon (clip) | 0.1 ~ 0.3 | Linear |
| Batch size | 32 ~ 512 | x2 steps |
| n_epochs | 1 ~ 20 | x2 steps |

推荐策略:
1. 先用默认值 (PPO: lr=3e-4, gamma=0.99, lambda=0.95, epsilon=0.2)
2. 调整 lr (最重要)
3. 调整 batch size 和 n_epochs
4. 微调 gamma 和 lambda
5. 最后调整 epsilon

## 4. 统计显著性

在比较两个算法时:
- 至少跑 5 个随机种子
- 使用 bootstrap confidence interval 而非简单均值+/-标准差
- 报告 interquartile mean (IQM) 而非简单均值 (对 outlier 更鲁棒)
- 不要只看最终性能, 也要看 sample efficiency (达到同样性能需要的步数)

## 5. 实验记录清单

每次实验应记录:
- 环境名称和版本
- 算法和所有超参数
- 随机种子
- 硬件 (CPU/GPU)
- 训练时间
- 最终性能 (mean +/- std over seeds)
- 主要发现
- 失败尝试 (与成功的同样重要!)

## 6. 练习

2. 对 lr 做参数扫描 (1e-4, 3e-4, 1e-3), 画对比曲线
3. 设计一个实验来验证 clipping 是否真的约束了策略更新
