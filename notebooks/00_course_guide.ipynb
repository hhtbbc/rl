
## 📖 学习方法建议

### 每个 Notebook 的使用方式

1. **先读**：通读 Markdown 中的直觉和数学推导
2. **手算**：用小例子手算验证关键公式
3. **运行**：逐步执行代码单元，理解每行代码
4. **修改**：尝试修改超参数、网络结构
5. **做题**：完成末尾的练习

### 重要原则

- **不要跳过数学**：每个公式都有存在的原因
- **不要只看代码**：理解背后的原理才算真正学会
- **不要只跑一次**：改变随机种子、超参数，观察结果变化
- **记录笔记**：用自己的话重新解释关键概念

### 预计时长

- 每天 2-4 小时
- 快速模式（理论为主）：每天约 2 小时
- 完整模式（含实验）：每天约 4 小时


## 💡 使用技巧

### 快速模式 vs 完整模式

每个算法 Notebook 包含两种训练配置：

- **快速模式** (`fast_mode=True`)：减少训练步数，2-3 分钟内完成，用于验证代码正确性
- **完整模式** (`fast_mode=False`)：完整训练配置，可能需要 10-30 分钟，用于获得可发表的结果

首次学习建议使用快速模式，理解算法后再用完整模式验证效果。

### GPU 使用

课程代码会自动检测 CUDA：
```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

所有算法都可以在 CPU 上运行（只是更慢）。如果 CUDA 可用，会自动加速。
